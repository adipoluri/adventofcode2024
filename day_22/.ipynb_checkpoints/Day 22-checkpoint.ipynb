```python
## Format Input
secret_numbers =  [int(row.strip()) for row in open('input.txt').readlines()]
test = [1,2,3,2024]
```

```python
## part 1
scrt = [[0 for num in range(2000)] for row in range(len(secret_numbers))]
sum = 0
for index, num in enumerate(secret_numbers):
    sec = num
    
    for i in range(2000):
        sec = ((sec * 64) ^ sec) % 16777216
        sec = ((sec // 32) ^ sec) % 16777216
        sec = ((sec * 2048) ^ sec) % 16777216
        scrt[index][i] = sec
    sum += sec
sum
```

```python
## part 2
cache = {}

def getPattern(monkey_num, index):
    vals = [scrt[monkey_num][index-i]%10 for i in reversed(range(5))]
    return [vals[i] - vals[i-1] for i in range(1,5)]

def getPriceForPattern(monkey_num, pattern):
    if (monkey_num, str(pattern)) in cache:
        return cache[(monkey_num, str(pattern))]

    for i in range(4,2000):
        curr = getPattern(monkey_num,i)
        if curr == pattern:
            cache[(monkey_num, str(pattern))] = scrt[monkey_num][i]
            return scrt[monkey_num][i]%10
            
    return 0

prices = []
for monkey_num in range(len(secret_numbers)):
    checked = []
    print("Analyzing Monkey: " + str(monkey_num))
    for i in range(4,2000):
        curr = getPattern(monkey_num,i)
        if curr in checked:
            continue

        price = 0
        for m in range(len(secret_numbers)):
            price += getPriceForPattern(monkey_num, curr)
            
        prices.append(price)
        checked.append(curr)

print(max(prices))
```
